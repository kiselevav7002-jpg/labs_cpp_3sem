```python
#task 0  
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
df = pd.read_csv(r"C:\Users\kisel\projects\helloworld\task0.csv")
df
plt.figure(figsize=(10, 6))
plt.plot(df["N"], df["bubble"],    marker="o", label="bubble")
plt.plot(df["N"], df["insertion"], marker="s", label="insertion")
plt.plot(df["N"], df["selection"], marker="^", label="selection")
plt.xlabel("N")
plt.ylabel("time, s")
plt.title("O(N^2) сортировки")
plt.legend()
plt.grid(True)
plt.show()

plt.figure(figsize=(10, 6))
plt.plot(np.log(df["N"]), np.log(df["bubble"]),    marker="o", label="bubble")
plt.plot(np.log(df["N"]), np.log(df["insertion"]), marker="s", label="insertion")
plt.plot(np.log(df["N"]), np.log(df["selection"]), marker="^", label="selection")
plt.xlabel("ln N")
plt.ylabel("ln t")
plt.title("O(N^2) Доказательство асимптотики: должно быть ~2")
plt.legend()
plt.grid(True)
plt.show()

for col in ["bubble", "insertion", "selection"]:
    slope = np.polyfit(np.log(df["N"]), np.log(df[col]), 1)[0]
    print(f"{col} slope = {slope:.3f}")
```

```python
#task2
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
df = pd.read_csv(r"C:\Users\kisel\projects\helloworld\task2.csv")
df
plt.figure(figsize=(10, 6))
plt.scatter(df["N"], df["quick"], marker="o", label="quick")
plt.scatter(df["N"], df["merge"], marker="s", label="merge")
plt.scatter(df["N"], df["heapq"], marker="^", label="heapq")
plt.xlabel("N")
plt.ylabel("time, s")
plt.title("O(NlogN) сортировки")
plt.legend()
plt.grid(True)
plt.show()

plt.figure(figsize=(10, 6))
plt.scatter(df["N"], df["quick"]/(df["N"]*np.log2(df["N"])),    marker="o", label="quick")
plt.scatter(df["N"], df["merge"]/(df["N"]*np.log2(df["N"])), marker="s", label="merge")
plt.scatter(df["N"], df["heapq"]/(df["N"]*np.log2(df["N"])), marker="^", label="heapq")
plt.xlabel("N")
plt.ylabel("t/(NlogN)")
plt.title("O(NlogN) проверка асимптотики")
plt.legend()
plt.grid(True)
plt.show()
for col in ["quick", "merge", "heapq"]:
    slope = np.polyfit((df["N"]), (df[col])/np.log2(df["N"]), 1)[0]
    slope=slope*(10**7)
    print(f"{col} slope = {slope:.3f}*10^(-7)")
```

```python
#task5
import time
import random
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
def bubble_sort(a):
  n = len(a)
  i = 0
  swapped = True
  while swapped:
    swapped = False
    for j in range(n-i-1):
      if a[j] > a[j+1]:
        a[j], a[j+1] = a[j+1], a[j]
        swapped = True
    i += 1

sizes = [1000, 2000, 4000, 8000, 16000]
times = [] 
for n in sizes:
    arr = [random.randint(0, 1000000) for _ in range(n)]

    start = time.perf_counter()
    bubble_sort(arr)
    end = time.perf_counter()
    times.append(end - start) 
print(times)


dfb = pd.read_csv(r"C:\Users\kisel\projects\helloworld\task5_bubble.csv")
plt.figure(figsize=(10, 6))
plt.plot(dfb["N"], dfb["bubble"], marker="o", label="C++")
plt.plot(sizes, times, marker="s", label="Python") 
plt.xlabel("N")
plt.ylabel("time, s")
plt.title("Сравнение python и C++")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import time
import random
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

def selection_sort(a):
  n = len(a)
  for i in range(n-1):
    k = i
    for j in range(i+1, n):
      if a[j] < a[k]:
        k = j
    a[k], a[i] = a[i], a[k]


sizes = [1000, 2000, 4000, 8000, 16000]
times = [] 
for n in sizes:
    arr = [random.randint(0, 1000000) for _ in range(n)]

    start = time.perf_counter()
    selection_sort(arr)
    end = time.perf_counter()
    times.append(end - start) 
print(times)



dfs= pd.read_csv(r"C:\Users\kisel\projects\helloworld\task5_selection.csv")
plt.figure(figsize=(10, 6))
plt.plot(dfs["N"], dfs["selection"], marker="o", label="C++")
plt.plot(sizes, times, marker="s", label="Python") 
plt.xlabel("N")
plt.ylabel("time, s")
plt.title("Сравнение python и C++")
plt.legend()
plt.grid(True)
plt.show()
```
