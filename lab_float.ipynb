```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv(r"C:\Users\kisel\projects\helloworld\lab_2_float\time_results.csv")
plt.figure(figsize=(10, 6))
formulas = df["formula"].unique()

for formula in formulas:
    sub_df = df[(df["formula"] == formula) & (df["seconds"] != -1)]
    
    if not sub_df.empty:
        plt.plot(sub_df["d"], sub_df["seconds"], marker="o", linewidth=2, label=formula)

# 5. Оформление графика
plt.xlabel("d, количество совпадающих знаков после запятой")
plt.ylabel("Среднее время, с")
plt.title("Время достижения точности для разных формул pi")
plt.xticks(range(0, 8))
plt.yscale("log")
plt.grid(True, which="major", linestyle="--", alpha=0.7)
plt.legend()
plt.savefig("time_graph.png", dpi=120)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv(r"C:\Users\kisel\projects\helloworld\lab_2_float\time_results.csv")
plt.figure(figsize=(10, 6))
formulas = df["formula"].unique()

for formula in formulas:

    sub_df = df[(df["formula"] == formula) & (df["iterations"] != -1)]
    
    if not sub_df.empty:
        plt.plot(sub_df["d"], sub_df["iterations"], marker="o", linewidth=2, label=formula)

plt.xlabel("d, знаков после запятой")
plt.ylabel("Количество итераций (N)")
plt.title("Зависимость числа итераций от требуемой точности")

plt.xticks(range(0, 8))
plt.yscale("log") 
plt.grid(True, which="major", linestyle="--", alpha=0.7)
plt.legend()
plt.savefig("iterations_graph.png", dpi=120)
plt.show()
```
